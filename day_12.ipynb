```typescript
// Preprocess data
const exampleRawData = `
0:
###
##.
##.

1:
###
##.
.##

2:
.##
###
##.

3:
##.
###
##.

4:
###
#..
###

5:
###
.#.
###

4x4: 0 0 0 0 2 0
12x5: 1 0 1 0 2 2
12x5: 1 0 1 0 3 2
`.slice(1, -1);

const rawData = await Deno.readTextFile("day_12.txt");

function preprocess(rawData: string) {
  const parts = rawData.split("\n\n");
  const sections = parts.pop()!.split("\n")
    .map(row => {
      const [width, height, ...nums] = [...row.match(/\d+/g)!];
      return [+width, +height, nums.map(Number)] as [number, number, number[]];
    });
  // const presents = parts.map(p => p.split("\n", 2).pop()!.split("").map(c => c === "#"));
  const presents = parts.map(p => p.split("\n", 2).pop()!.split("").map(c => c === "#").filter(s => s).length);
  return [presents, sections] as const;
}
```

```typescript
function part1([presents, sections]: ReturnType<typeof preprocess>) {
  let total = 0;
  for (const [width, height, indexes] of sections) {
    const area = new Array(width * height).fill(false);
    const indexesFlat = indexes.flatMap((e, i) => new Array(e).fill(i));
    function fitPresents(i: number, area: boolean[]) {
      if (i === indexesFlat.length) return true;
      let present = presents[indexesFlat[i]];
      for (const flip of [false, true]) {
        if (flip) {
          present = [
            present[2], present[1], present[0],
            present[5], present[4], present[3],
            present[8], present[7], present[6],
          ];
        }
        for (let y = 0; y < height - 2; y++) {
          for (let x = 0; x < width - 2; x++) {
            // 0 deg
            x: {
              const clone = [...area];
              if (present[0] && clone[y * width + x]) break x;
              if (present[1] && clone[y * width + x + 1]) break x;
              if (present[2] && clone[y * width + x + 2]) break x;
              if (present[3] && clone[(y + 1) * width + x]) break x;
              if (present[4] && clone[(y + 1) * width + x + 1]) break x;
              if (present[5] && clone[(y + 1) * width + x + 2]) break x;
              if (present[6] && clone[(y + 2) * width + x]) break x;
              if (present[7] && clone[(y + 2) * width + x + 1]) break x;
              if (present[8] && clone[(y + 2) * width + x + 2]) break x;
              clone[y * width + x] ||= present[0];
              clone[y * width + x + 1] ||= present[1];
              clone[y * width + x + 2] ||= present[2];
              clone[(y + 1) * width + x] ||= present[3];
              clone[(y + 1) * width + x + 1] ||= present[4];
              clone[(y + 1) * width + x + 2] ||= present[5];
              clone[(y + 2) * width + x] ||= present[6];
              clone[(y + 2) * width + x + 1] ||= present[7];
              clone[(y + 2) * width + x + 2] ||= present[8];
              if (fitPresents(i + 1, clone)) return true;
            }
            // 90 deg
            x: {
              const clone = [...area];
              if (present[2] && clone[y * width + x]) break x;
              if (present[5] && clone[y * width + x + 1]) break x;
              if (present[8] && clone[y * width + x + 2]) break x;
              if (present[1] && clone[(y + 1) * width + x]) break x;
              if (present[4] && clone[(y + 1) * width + x + 1]) break x;
              if (present[7] && clone[(y + 1) * width + x + 2]) break x;
              if (present[0] && clone[(y + 2) * width + x]) break x;
              if (present[3] && clone[(y + 2) * width + x + 1]) break x;
              if (present[6] && clone[(y + 2) * width + x + 2]) break x;
              clone[y * width + x] ||= present[2];
              clone[y * width + x + 1] ||= present[5];
              clone[y * width + x + 2] ||= present[8];
              clone[(y + 1) * width + x] ||= present[1];
              clone[(y + 1) * width + x + 1] ||= present[4];
              clone[(y + 1) * width + x + 2] ||= present[7];
              clone[(y + 2) * width + x] ||= present[0];
              clone[(y + 2) * width + x + 1] ||= present[3];
              clone[(y + 2) * width + x + 2] ||= present[6];
              if (fitPresents(i + 1, clone)) return true;
            }
            // 180 deg
            x: {
              const clone = [...area];
              if (present[8] && clone[y * width + x]) break x;
              if (present[7] && clone[y * width + x + 1]) break x;
              if (present[6] && clone[y * width + x + 2]) break x;
              if (present[5] && clone[(y + 1) * width + x]) break x;
              if (present[4] && clone[(y + 1) * width + x + 1]) break x;
              if (present[3] && clone[(y + 1) * width + x + 2]) break x;
              if (present[2] && clone[(y + 2) * width + x]) break x;
              if (present[1] && clone[(y + 2) * width + x + 1]) break x;
              if (present[0] && clone[(y + 2) * width + x + 2]) break x;
              clone[y * width + x] ||= present[8];
              clone[y * width + x + 1] ||= present[7];
              clone[y * width + x + 2] ||= present[6];
              clone[(y + 1) * width + x] ||= present[5];
              clone[(y + 1) * width + x + 1] ||= present[4];
              clone[(y + 1) * width + x + 2] ||= present[3];
              clone[(y + 2) * width + x] ||= present[2];
              clone[(y + 2) * width + x + 1] ||= present[1];
              clone[(y + 2) * width + x + 2] ||= present[0];
              if (fitPresents(i + 1, clone)) return true;
            }
            // 270 deg
            x: {
              const clone = [...area];
              if (present[6] && clone[y * width + x]) break x;
              if (present[3] && clone[y * width + x + 1]) break x;
              if (present[0] && clone[y * width + x + 2]) break x;
              if (present[7] && clone[(y + 1) * width + x]) break x;
              if (present[4] && clone[(y + 1) * width + x + 1]) break x;
              if (present[1] && clone[(y + 1) * width + x + 2]) break x;
              if (present[8] && clone[(y + 2) * width + x]) break x;
              if (present[5] && clone[(y + 2) * width + x + 1]) break x;
              if (present[2] && clone[(y + 2) * width + x + 2]) break x;
              clone[y * width + x] ||= present[6];
              clone[y * width + x + 1] ||= present[3];
              clone[y * width + x + 2] ||= present[0];
              clone[(y + 1) * width + x] ||= present[7];
              clone[(y + 1) * width + x + 1] ||= present[4];
              clone[(y + 1) * width + x + 2] ||= present[1];
              clone[(y + 2) * width + x] ||= present[8];
              clone[(y + 2) * width + x + 1] ||= present[5];
              clone[(y + 2) * width + x + 2] ||= present[2];
              if (fitPresents(i + 1, clone)) return true;
            }
          }
        }
      }
      return false;
    }
    if (fitPresents(0, [...area])) total += 1;
  }
  return total;
}

part1(
  preprocess(
    // exampleRawData,
    rawData,
  )
);
```
